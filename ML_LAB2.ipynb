###**SLR**

```python
x=df[['Avg. Area Income']]
y=df['Price']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=101)
```

```python
from sklearn.linear_model import LinearRegression
slr_model=LinearRegression()
slr_model.fit(x_train, y_train)
```

```python
y_pred=slr_model.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
import numpy as np

print("Simple Linear Regression")
print("------------------------")
print("Mean Absolute Error:", mean_absolute_error(y_test, y_pred))
print("Mean Squared Error:", mean_squared_error(y_test, y_pred))
print("Root Mean Squared Error:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R2 Score:", r2_score(y_test, y_pred))
```

```python
print("Intercept:",slr_model.intercept_)
print("Slope:",slr_model.coef_[0])
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(x_test,y_test,color='blue',label='Actual Data')
plt.plot(x_test,y_pred,color='red',linewidth=2,label='Regression Line')
plt.xlabel('Average Area Income')
plt.ylabel('Price')
plt.title('Simple Linear Regression')
plt.legend()
plt.show()
```

##**Ridge Regression**

Hyper Parameter

```python
from sklearn.model_selection import GridSearchCV
```

```python
from sklearn.linear_model import Ridge

ridge = Ridge()

ridge.fit(x_train, y_train)

ridge_pred = ridge.predict(x_test)

print("Default Ridge Regression:", r2_score(y_test, ridge_pred))
```

```python
param_grid = {'alpha': [0.001, 0.01, 0.1, 1, 10, 100]}
```

```python
grid_ridge=GridSearchCV(estimator=Ridge(),param_grid=param_grid,cv=5,scoring='r2')
```

```python
grid_ridge.fit(x_train,y_train)
```

```python
print("Best Alpha:",grid_ridge.best_params_)
print("Best Score:",grid_ridge.best_score_)
```

```python
best_ridge=grid_ridge.best_estimator_
ridge_pred=best_ridge.predict(x_test)
```
